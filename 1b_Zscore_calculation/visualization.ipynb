```python
import pandas as pd
from plotnine import ggplot, aes, geom_density, labs, geom_vline
```

```python
df = pd.read_csv('clean/test_data.csv')
ggplot(df) + geom_density(aes(x='C1 Clutch')) + labs(title='Distribution of C1 Clutch test pass rates')


```

```python
#df2 = pd.read_excel('clean/CLEANED_Full_Claims_Report.xlsx', Page=2)
df2 = pd.read_csv("clean/test_data_zscore.csv")#, header=0) 
df2 = df2[df2['Test'] == 'C1 Clutch']
```

```python
ggplot(df2) + geom_density(aes(x='z_score')) + \
    geom_vline(aes(xintercept=2), color='red') + \
    geom_vline(aes(xintercept=-2), color='red') + \
    labs(title='Distribution of C1 CLUTCH zscores')
```
